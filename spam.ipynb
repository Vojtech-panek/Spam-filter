```python
import io
import tarfile
import urllib.request
from pathlib import Path

BASE_URL = "https://spamassassin.apache.org/old/publiccorpus/"

DATASETS = [
    "20030228_easy_ham.tar.bz2",
     "20030228_spam.tar.bz2"
]


def download_and_extract_datasets(
    base_url: str,
    filenames: list[str],
) -> list[Path]:
    """Download and extract datasets if they don't exist, returning their paths."""
    datasets_dir = Path.cwd() / "datasets"
    datasets_dir.mkdir(parents=True, exist_ok=True)

    headers = {"User-Agent": "Mozilla/5.0"}
    extracted_paths = []

    for filename in filenames:
        # Determine the target directory name based on the archive name
        dir_name = "easy_ham" if "ham" in filename else "spam"
        target_dir = datasets_dir / dir_name
        extracted_paths.append(target_dir)

        # Skip downloading and extracting if the target directory already exists
        if target_dir.is_dir():
            print(f"Directory '{dir_name}' already exists, skipping download.")
            continue

        url = base_url + filename
        request = urllib.request.Request(url, headers=headers)

        print(f"Downloading and extracting: {filename}...")

        with urllib.request.urlopen(request) as response:
            content = response.read()

        with tarfile.open(
            fileobj=io.BytesIO(content),
            mode="r:*",
        ) as archive:
            archive.extractall(datasets_dir, filter="data")

    print("Done!")
    return sorted(extracted_paths)




  
```

```python
ham_dir, spam_dir = download_and_extract_datasets(BASE_URL, DATASETS)
```

```python
ham_filenames = [f for f in sorted(ham_dir.iterdir()) if len(f.name) > 20]
spam_filenames = [f for f in sorted(spam_dir.iterdir()) if len(f.name) > 20]
```

```python
import email
import email.policy

def load_email(filepath):
    with open(filepath, "rb") as f:
        return email.parser.BytesParser(policy=email.policy.default).parse(f)
```

```python
ham_emails = [load_email(filepath) for filepath in ham_filenames]
spam_emails = [load_email(filepath) for filepath in spam_filenames]
```

```python
from email.message import Message
from typing import Union

def get_email_structure(email: Union[str, Message]) -> str:
    if isinstance(email, str):
        return email
        
    payload = email.get_payload()
    
    if isinstance(payload, list):
        
        multipart = ", ".join(get_email_structure(sub) for sub in payload)
        return f"multipart({multipart})"
        
    # Zbytečné 'else' bylo odstraněno
    return email.get_content_type()
```

```python
from collections import Counter 
def structures_counter(emails):
    return Counter(get_email_structure(email) for email in emails)
```

```python
import re
from html import unescape

def html_to_plain_text(html):
    text = re.sub('<head.*?>.*?</head>', '', html, flags=re.M | re.S | re.I)
    text = re.sub('<a\s.*?>', ' HYPERLINK ', text, flags=re.M | re.S | re.I)
    text = re.sub('<.*?>', '', text, flags=re.M | re.S)
    text = re.sub(r'(\s*\n)+', '\n', text, flags=re.M | re.S)
    return unescape(text)
```

```python

from bs4 import BeautifulSoup


def html_to_plain_text1(html):
    soup = BeautifulSoup(html, "html.parser")

    # <head>...</head> → ''
    if soup.head:
        soup.head.decompose()

    # <a ...> → ' HYPERLINK '
    for link in soup.find_all("a"):
        link.replace_with(" HYPERLINK " + link.get_text())

    # <tag> → ''
    text = soup.get_text()

    # (\s*\n)+ → '\n'
    lines = [line for line in text.splitlines() if line.strip()]
    text = "\n".join(lines)

    # HTML entity → normální znak
    return unescape(text)
```
